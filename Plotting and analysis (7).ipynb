# Solar System Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt

# Path to your CSV file
file_path = r"C:\Users\Abdullah\dispatch_outputs_solar_dc\dispatch_data_2024.csv"

# Load the CSV into a DataFrame
df = pd.read_csv(file_path)

# Plot each column against Hour
plt.figure(figsize=(14, 6))
plt.plot(df['Hour'], df['Load (kW)'], label='Load (kW)')
plt.plot(df['Hour'], df['Solar (kW)'], label='Solar (kW)')
plt.plot(df['Hour'], df['Wind (kW)'], label='Wind (kW)')
plt.plot(df['Hour'], df['SOC (kWh)'], label='SOC (kWh)')

plt.xlabel('Hour')
plt.ylabel('Power / Energy')
plt.title('Load, Solar, Wind, and SOC over Time')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Path to your CSV file
file_path = r"C:\Users\Abdullah\dispatch_outputs_solar-wind_dc_80metLoadCostCurtailment\dispatch_data_2024.csv"

# Load the CSV into a DataFrame
df = pd.read_csv(file_path)

# Plot each column against Hour
plt.figure(figsize=(14, 6))
plt.plot(df['Hour'], df['Load (kW)'], label='Load (kW)')
plt.plot(df['Hour'], df['Solar (kW)'], label='Solar (kW)')
plt.plot(df['Hour'], df['Wind (kW)'], label='Wind (kW)')
plt.plot(df['Hour'], df['SOC (kWh)'], label='SOC (kWh)', alpha=0.5)  # More transparent

plt.xlabel('Hour')
plt.ylabel('Power / Energy')
plt.title('The 2024 Capacity Load for 80% Met Load Sizing, Solar, Wind and SOC in 2024')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
